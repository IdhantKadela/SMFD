**Buffon’s Needle — Monte Carlo Estimation of π**

```python
import numpy as np
import matplotlib.pyplot as plt

# Buffon's Needle Simulation to Estimate π

# Parameters
N = 10000             # Total number of needle drops
d = 1.0               # Distance between the lines
l = 1.0               # Length of the needle (must be <= d)

# Initialization
crosses = 0           # Counter for how many times needle crosses a line
estimates = []        # Store intermediate estimates of π for convergence plot

# Optional: set random seed for reproducibility
np.random.seed(0)

# Simulation loop
for n in range(1, N + 1):
    x = np.random.uniform(0, d / 2)         # Distance from center to nearest line
    theta = np.random.uniform(0, np.pi / 2) # Angle with horizontal

    # Check if needle crosses a line
    if (l / 2) * np.sin(theta) >= x:
        crosses += 1

    # Estimate π only after at least one crossing
    if crosses > 0:
        pi_estimate = (2 * n) / crosses
        estimates.append(pi_estimate)
    else:
        estimates.append(np.nan)

# Final estimated value of π
print(f"Estimated value of π after {N} drops: {estimates[-1]:.6f}")

# --------------------------------------
# Plotting convergence
# --------------------------------------
plt.figure(figsize=(10, 6))
plt.plot(np.log10(np.arange(1, N + 1)), estimates, label='Estimated π')
plt.axhline(np.pi, color='red', linestyle='--', label='True π ≈ 3.14159')
plt.xlabel("log₁₀(Number of Drops)")
plt.ylabel("Estimated π")
plt.title("Convergence of Estimated π using Buffon’s Needle")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
